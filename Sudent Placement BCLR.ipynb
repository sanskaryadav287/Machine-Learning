### Remove outliers

```python
Q1=df['package_lpa'].quantile(0.25)
Q3=df['package_lpa'].quantile(0.75)

IQR = Q3-Q1
```

```python
upper_limit=Q3+1.5*IQR
lower_limit=Q1-1.5*IQR
```

```python
df['package_lpa']=np.where(df['package_lpa']> upper_limit,upper_limit,np.where(df['package_lpa']<lower_limit,lower_limit,df['package_lpa']))
```
